```python
# 관련성 청크 삭제
threshold = 0.8

filtered = [
    c for c in chunks
    if c["score"] >= threshold
]

print(filtered)
```

```python
# 중복 내용 제거
unique = []
seen = set()

for c in filtered:
    if c["text"] not in seen:
        unique.append(c)
        seen.add(c["text"])
print(seen)
# 혹은 cosine_similarity() <- 이걸로 유사도가 일정 수치 이상이면 제거하게끔도 가능
```

```python
# 문서 개수 제한

from collections import defaultdict

limit = 2

result = []
counter = defaultdict(int)

for c in unique:
    if counter[c["source"]] < limit:
        result.append(c)
        counter[c["source"]] += 1
print(result)
```

```python
# 문서 그룹마다 원하는 내용 보기

from collections import defaultdict

groups = defaultdict(list)

for c in chunks:
    groups[c["source"]].append(c['text'])

print(groups)
```

```python
# 연차에 대한 최신 정보만 필요할때

recent = {}

for c in chunks:
    source = c["source"]

    if source not in recent:
        recent[source] = c

    elif c['source'] == recent[source]['source'] and c["year"] > recent[source]["year"]:
        recent[source] = c

print(list(recent.values()))
```
